```python
# While loop while loop is used for where we don't know where to stop but we have conditions . it can run till infinity
lucky_number = 5675
while True:
  number = int(input("Enter a lucky number:"))
  if number >=1000 and number <= 9999:
    if number == lucky_number:
      print("Congratulations you have guessed the number")
      break
    else:
      print("try again")
  else:
    print("Enter any 4-digit number")
```

```python
# take two inputs from the user and if the product is greater then 5000 print pass else keep trying
while True:
  num1 = int(input("Enter a number:"))
  num2 = int(input("Enter a number:"))
  if num1 > 0 and num2 > 0:
    if num1 * num2 > 5000:
      print("pass")
      break
    else:
      print("Keep trying")
  else:
    print("please enter a valid number")
```
